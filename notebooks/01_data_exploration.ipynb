```python
# Quick exploration in Python directly (or use Jupyter)
import pandas as pd
import matplotlib.pyplot as plt

# Load your data
df = pd.read_csv("data/blockchain_ticketing_master.csv")
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst 5 rows:")
print(df.head())

# Check target distribution
if 'scalping_label' in df.columns:
    plt.figure(figsize=(10, 4))
    df['scalping_label'].value_counts().plot(kind='bar')
    plt.title('Scalping Label Distribution')
    plt.show()
```
